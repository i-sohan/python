NumPy (short for Numerical Python) is the foundational open-source library for numerical and scientific computing in Python. It provides support for large, multi-dimensional arrays and matrices, along with a collection of high-performance mathematical functions to operate on them efficiently.

Nearly the entire Python data science ecosystem—including Pandas, SciPy, Scikit-learn, PyTorch, and TensorFlow—is built on top of NumPy.
